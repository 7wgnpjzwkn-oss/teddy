```
import os
import json
import math
import bpy
import mathutils

from mathutils import Vector
from math import radians, sin, cos


# ==========================================
# 1. CONFIGURATION AND INITIALIZATION
# ==========================================

BASE_DIR = "/Users/dobbinp1/Downloads"
CONFIG_FILE = os.path.join(BASE_DIR, "teddy_config.json")
IK_FILE = os.path.join(BASE_DIR, "teddy_ik_defaults.json")
OUTPUT_VIDEO_PATH = os.path.join(BASE_DIR, "teddy_pose_transition.mp4")

START_FRAME = 1
#FRAMES_PER_TRANSITION = 60
#HOLD_FRAMES = 12


from mathutils import Vector

from mathutils import Vector

class LegDefinition:

    def __init__(self):

        #
        # Rest pose joint positions
        #
        self.hip = Vector((0,0,0))
        self.knee = Vector((0,0,0))
        self.ankle = Vector((0,0,0))

        #
        # Derived measurements
        #
        self.upper_length = 0.0
        self.lower_length = 0.0

class LegRigSolution:

    def __init__(self):

        self.target = None
        self.pole = None


class LegGeometry:

    def __init__(self):

        self.hip = Vector((0, 0, 0))
        self.knee = Vector((0, 0, 0))
        self.ankle = Vector((0, 0, 0))
        self.thigh_angle =  Vector((0, 0, 0))
        self.shin_angle =  Vector((0, 0, 0))
        self.thigh_direction = Vector((0,0,0))
        self.shin_direction = Vector((0,0,0))





class LimbControls:

    def __init__(self):

        #
        # Current IK controls
        #
        self.target = None
        self.pole = None

        #
        # Anatomical pose
        #
        self.thigh_pitch = None
        self.knee_bend = None
        self.ankle_pitch = None




class RigControls:

    def __init__(self):

        self.right_leg = LimbControls()
        self.left_leg = LimbControls()

        self.right_arm = LimbControls()
        self.left_arm = LimbControls()

        self.torso = None



def build_leg_definition(
    armature_obj,
    side
):

    leg = LegDefinition()

    upper = armature_obj.data.bones[f"Mch.Leg.Upper.{side}"]
    lower = armature_obj.data.bones[f"Mch.Leg.Lower.{side}"]

    #
    # Rest joint positions
    #

    leg.hip = armature_obj.matrix_world @ upper.head_local

    leg.knee = armature_obj.matrix_world @ upper.tail_local

    leg.ankle = armature_obj.matrix_world @ lower.tail_local

    #
    # Bone lengths
    #

    leg.upper_length = (
        leg.knee - leg.hip
    ).length

    leg.lower_length = (
        leg.ankle - leg.knee
    ).length

    return leg

def print_leg_definition(
    side,
    leg
):

    print(f"\n{side}")

    print(f"Hip   : {leg.hip}")

    print(f"Knee  : {leg.knee}")

    print(f"Ankle : {leg.ankle}")

    print(f"Upper Length : {leg.upper_length:.4f}")

    print(f"Lower Length : {leg.lower_length:.4f}")

def geometry_to_leg_rig(
    geometry,
    side
):
    """
    Converts anatomical geometry into the controls
    expected by our IK rig.
    """

    rig = LegRigSolution()

    #
    # IK target is simply the ankle.
    #

    rig.target = geometry.ankle.copy()

    #
    # Temporary pole.
    #

    pole_offset = 0.25

    if side == "R":
        rig.pole = geometry.knee + Vector((pole_offset,0,0))
    else:
        rig.pole = geometry.knee + Vector((-pole_offset,0,0))

    return rig


def solve_leg_geometry(
    leg_definition,
    thigh_pitch,
    knee_bend
):
    """
    Calculates the knee and ankle positions from anatomical angles.

    Returns a LegGeometry.
    """

    geometry = LegGeometry()

    geometry.hip = leg_definition.hip.copy()

    #
    # Upper leg
    #

    #upper_angle = radians(upper_pitch)

    thigh_angle = radians(thigh_pitch)

    geometry.thigh_angle = thigh_angle

    knee = hip + Vector((
        0.0,
        leg_definition.upper_length * sin(thigh_angle),
        -leg_definition.upper_length * cos(thigh_angle),
    ))

    geometry.knee = knee

    #
    # Lower leg
    #

    shin_angle = radians(thigh_pitch - knee_bend)

    geometry.shin_angle = shin_angle

    ankle = knee + Vector((
        0.0,
        leg_definition.lower_length * sin(shin_angle),
        -leg_definition.lower_length * cos(shin_angle),
    ))

    geometry.ankle = ankle

    geometry.thigh_direction = (
        geometry.knee - geometry.hip
    ).normalized()

    geometry.shin_direction = (
        geometry.ankle - geometry.knee
    ).normalized()

    return geometry

def apply_leg_controls(
    armature_obj,
    side,
    leg_controls,
    leg_definition
):

    #
    # New anatomical path
    #

    if leg_controls.thigh_pitch is not None:

        geometry = solve_leg_geometry(

            leg_definition,

            thigh_pitch=leg_controls.thigh_pitch,

            knee_bend=leg_controls.knee_bend,
        )

        rig = geometry_to_leg_rig(
            geometry,
            side
        )

        apply_world_location(
            armature_obj,
            f"Leg.Target.{side}",
            rig.target
        )

        apply_world_location(
            armature_obj,
            f"Leg.Pole.{side}",
            rig.pole
        )

        return

    #
    # Existing path
    #

    apply_world_location(
        armature_obj,
        f"Leg.Target.{side}",
        leg_controls.target
    )

    apply_world_location(
        armature_obj,
        f"Leg.Pole.{side}",
        leg_controls.pole
    )



def apply_arm_controls(
    armature_obj,
    side,
    arm_controls
):

    apply_world_location(
        armature_obj,
        f"Arm.Target.{side}",
        arm_controls.target
    )

    apply_world_location(
        armature_obj,
        f"Arm.Pole.{side}",
        arm_controls.pole
    )


def apply_torso_controls(
    armature_obj,
    controls
):

    apply_world_location(
        armature_obj,
        "Torso",
        controls.torso
    )

def build_rig_controls(pose_data):

    controls = RigControls()

    targets = pose_data["targets"]

    controls.torso = targets.get("Torso")

    controls.right_leg.target = targets.get("Leg.Target.R")
    controls.left_leg.target  = targets.get("Leg.Target.L")

    controls.right_leg.pole = targets.get("Leg.Pole.R")
    controls.left_leg.pole  = targets.get("Leg.Pole.L")

    controls.right_arm.target = targets.get("Arm.Target.R")
    controls.left_arm.target  = targets.get("Arm.Target.L")

    controls.right_arm.pole = targets.get("Arm.Pole.R")
    controls.left_arm.pole  = targets.get("Arm.Pole.L")

    right_leg = pose_data.get("right_leg", {})

    print(f"Right Leg: {right_leg}")

    controls.right_leg.thigh_pitch = right_leg.get("thigh_pitch")
    controls.right_leg.knee_bend = right_leg.get("knee_bend")
    controls.right_leg.ankle_pitch = right_leg.get("ankle_pitch")

    return controls

def apply_world_location(armature_obj, bone_name, coord):

    if coord is None:
        return

    p_bone = armature_obj.pose.bones.get(bone_name)

    if p_bone is None:
        return

    global_target_vec = mathutils.Vector(coord)

    matrix_translation = mathutils.Matrix.Translation(global_target_vec)

    local_matrix = armature_obj.convert_space(
        pose_bone=p_bone,
        matrix=matrix_translation,
        from_space='WORLD',
        to_space='LOCAL'
    )

    local_loc = local_matrix.to_translation()

    print(
        f"  - Bone: {bone_name:15}"
        f" | Input World: [{coord[0]:.2f}, {coord[1]:.2f}, {coord[2]:.2f}]"
        f" -> Computed Local: "
        f"[{local_loc.x:.4f}, {local_loc.y:.4f}, {local_loc.z:.4f}]"
    )

    p_bone.location = local_loc
    p_bone.rotation_mode = 'QUATERNION'
    p_bone.rotation_quaternion = (1,0,0,0)


def keyframe_bone(armature_obj, bone_name, frame):

    p_bone = armature_obj.pose.bones.get(bone_name)

    if p_bone is None:
        return

    p_bone.keyframe_insert(
        data_path="location",
        frame=frame
    )

    p_bone.keyframe_insert(
        data_path="rotation_quaternion",
        frame=frame
    )

def get_cfg(acfg, path, default=None):
    """
    Existing function can be replaced with this version if desired.
    Supports dotted paths.
    """

    current = acfg

    for part in path.split("."):
        if isinstance(current, dict) and part in current:
            current = current[part]
        else:
            return default

    return current


def get_constant(cfg, name, default=None):
    return get_cfg(cfg, f"constants.{name}", default)


def get_shape(cfg, name):
    shape = get_cfg(cfg, f"shape_library.{name}")

    if shape is None:
        raise KeyError(f"Unknown shape '{name}'")

    return shape


def get_vector(cfg, name):
    vec = get_cfg(cfg, f"vector_library.{name}")

    if vec is None:
        raise KeyError(f"Unknown vector '{name}'")

    return mathutils.Vector(vec)


def get_offset(cfg, name, scale=1.0):

    offset = get_cfg(cfg, f"offset_library.{name}")

    if offset is None:
        raise KeyError(
            f"Unknown offset '{name}'"
        )

    return mathutils.Vector((
        offset[0] * scale,
        offset[1] * scale,
        offset[2] * scale
    ))


def get_rotation(cfg, name):
    value = get_cfg(cfg, f"rotation_library.{name}")

    if value is None:
        raise KeyError(f"Unknown rotation '{name}'")

    return math.radians(value)


def make_scaled_vector(cfg, vector_name, scale):
    vec = get_vector(cfg, vector_name)

    return mathutils.Vector(
        (
            vec.x * scale,
            vec.y * scale,
            vec.z * scale,
        )
    )


def make_scaled_offset(cfg, offset_name, scale):
    return get_offset(cfg, offset_name, scale)


def get_radius(cfg, shape_name, axis):
    shape = get_shape(cfg, shape_name)

    return shape["radius"][axis]


def build_shape_dimensions(cfg, shape_name, scale):
    shape = get_shape(cfg, shape_name)

    return (
        shape["radius"]["x"] * scale,
        shape["radius"]["y"] * scale,
        shape["radius"]["z"] * scale,
    )

def load_or_create_config():
    default_config = {
        "origin": [0.0, 0.0, 0.0],
        "base_unit_head_radius": 0.45,
        "constants": {},
        "vector_library": {},
        "offset_library": {},
        "rotation_library": {},
        "shape_library": {},
        "proportions": {
            "torso_height_factor": 2.2,
            "torso_width_factor": 0.7,
            "arm_length_factor": 1.1,
            "arm_thick_shoulder": 0.4,
            "arm_thick_paw": 0.25,
            "leg_length_factor": 1.7,
            "leg_thick": 0.5,
            "ear_angle_deg": 60.0,
            "ear_radius_factor": 0.35
        },
        "shape_exponents": {
            "default_sphere_n": 2.0,
            "snout_boxiness_n": 2.5,
            "torso_taper_factor": 0.3
        },
        "surface_deformation": {
            "use_plush_displacement": True,
            "noise_scale": 0.03,
            "noise_frequency": 25.0
        },
        "render_settings": {
            "resolution_x": 1080,
            "resolution_y": 1080,
            "samples": 32,
            "engine": "BLENDER_EEVEE"  # Correct standard designation for Blender 5.x+ "BLENDER_EEVEE"
        },
        "constants": {
            "upper_limb_root_extension": 0.20,
            "target_display_length": 0.10,
            "ear_distance": 1.00,
            "ear_rear_offset": 0.05,
            "snout_forward": 0.85,
            "snout_drop": 0.10,
            "nose_forward": 1.20,
            "nose_raise": 0.08
        },
        "animation": {
          "fps": 24,
          "hold_frames": 12,
          "transition_frames": 60
        },
        "vector_library": {

            "zero": [0.0, 0.0, 0.0],

            "up": [0.0, 0.0, 1.0],
            "down": [0.0, 0.0, -1.0],

            "forward": [0.0, 1.0, 0.0],
            "back": [0.0, -1.0, 0.0],

            "right": [1.0, 0.0, 0.0],
            "left": [-1.0, 0.0, 0.0],

            "arm_upper_right": [0.500, 0.000, -0.866],
            "arm_upper_left": [-0.500, 0.000, -0.866],

            "arm_lower_right": [0.200, 0.000, -0.980],
            "arm_lower_left": [-0.200, 0.000, -0.980],

            "leg_upper": [0.000, 0.000, -1.000],
            "leg_lower": [0.000, 0.000, -1.000],

            "arm_pole": [0.000, 0.500, 0.000],
            "leg_pole": [0.000, -0.500, 0.000]
        },
        "offset_library": {
            "target_tail": [0.0, 0.10, 0.0],
            "pole_tail": [0.0, 0.0, 0.10],
            "snout": [0.0, -0.85, -0.10],
            "nose": [0.0, -1.20, 0.08]
        },
        "rotation_library": {
            "foot_forward": 90.0,
            "ear_default": 60.0
        },
        "shape_library": {
            "head": {
                "radius": {
                    "x": 1.10,
                    "y": 1.00,
                    "z": 1.00
                }
            },
            "snout": {

                "radius": {
                    "x": 0.45,
                    "y": 0.35,
                    "z": 0.30
                },
                "shape_exponent": 2.5
            },
            "nose": {

                "radius": {
                    "x": 0.15,
                    "y": 0.10,
                    "z": 0.08
                },
                "shape_exponent": 1.8
            },
            "ear": {

                "radius": {
                    "x": 1.00,
                    "y": 0.30,
                    "z": 1.00
                }
            }
        }
    }
    os.makedirs(BASE_DIR, exist_ok=True)
    if os.path.exists(CONFIG_FILE):
        try:
            with open(CONFIG_FILE, "r") as f:
                return json.load(f)
        except Exception:
            pass
    with open(CONFIG_FILE, "w") as f:
        json.dump(default_config, f, indent=4)
    return default_config

def load_or_create_ik_defaults():
    default_ik = {
        "ik.default": {
            "chain_count": 2,
            "ik_linear_weight": 1.0,
            "ik_rotation_weight": 0.0,
            "iterations": 50,
            "pole_angle_deg": 0.0,
            "use_rotation": False
        },

        # --- GENERIC DATA-DRIVEN LIMB STRUCTURE MATRIX ---
        # Decouples structural layout parameters completely from generation code loops
        "limb_matrix": {
            "Arm.R": {
                "prefix": "Arm",
                "side_sign": 1,
                "suffix": "R",
                "parent_bone": "Torso",
                "deform_parent": "Torso",
                "mechanical_parent": "Torso",
                "target_parent": "Torso",
                "pole_parent": "Torso",
                "ik_space": "WORLD",
                "constraint_space": "WORLD",
                "origin": {
                    "region": "shoulder",
                    "x_factor": 0.85,
                    "z_factor": 0.75
                },
                "segment_length_factor": 1.0,
                "upper_vector": "arm_upper_right",
                "lower_vector": "arm_lower_right",
                "pole_vector": "arm_pole",
                "preferred_ik_angle": {
                    "x": 0.1,
                    "y": 0.0,
                    "z": 0.0
                },
                "pole_angle_deg": 0.0,
                "use_deform": True,
                "joint_limits": {
                    "upper": {
                        "use_limit_x": True,
                        "min_x": -0.785,
                        "max_x": 1.571,

                        "use_limit_y": True,
                        "min_y": -0.523,
                        "max_y": 0.523,

                        "use_limit_z": True,
                        "min_z": -0.785,
                        "max_z": 1.571
                    },
                    "lower": {
                        "use_limit_x": True,
                        "min_x": 0.0,
                        "max_x": 2.094
                    }
                }
            },
            "Arm.L": {
                "prefix": "Arm",
                "side_sign": -1,
                "suffix": "L",

                "parent_bone": "Torso",
                "deform_parent": "Torso",
                "mechanical_parent": "Torso",
                "target_parent": "Torso",
                "pole_parent": "Torso",

                "ik_space": "WORLD",
                "constraint_space": "WORLD",

                "origin": {
                    "region": "shoulder",
                    "x_factor": 0.85,
                    "z_factor": 0.75
                },

                "segment_length_factor": 1.0,

                "upper_vector":  "arm_upper_left",
                "lower_vector": "arm_lower_left",

                "pole_vector": "arm_pole",

                "preferred_ik_angle": {
                    "x": 0.1,
                    "y": 0.0,
                    "z": 0.0
                },

                "pole_angle_deg": 0.0,

                "use_deform": True,

                "joint_limits": {
                    "upper": {
                        "use_limit_x": True,
                        "min_x": -0.785,
                        "max_x": 1.571,

                        "use_limit_y": True,
                        "min_y": -0.523,
                        "max_y": 0.523,

                        "use_limit_z": True,
                        "min_z": -1.571,
                        "max_z": 0.785
                    },

                    "lower": {
                        "use_limit_x": True,
                        "min_x": 0.0,
                        "max_x": 2.094
                    }
                }
            },

            "Leg.R": {
                "prefix": "Leg",
                "side_sign": 1,
                "suffix": "R",

                "parent_bone": "Torso",
                "deform_parent": "Torso",
                "mechanical_parent": "Torso",
                "target_parent": "Torso",
                "pole_parent": "Torso",

                "ik_space": "WORLD",
                "constraint_space": "WORLD",

                "origin": {
                    "region": "hip",
                    "x_factor": 0.65,
                    "z_factor": 0.15
                },

                "segment_length_factor": 1.0,

                "upper_vector": "leg_upper",
                "lower_vector": "leg_lower",

                "pole_vector": "leg_pole",

                "preferred_ik_angle": {
                    "x": -0.1,
                    "y": 0.0,
                    "z": 0.0
                },

                "pole_angle_deg": 0.0,

                "align_roll_axis": [1.0, 0.0, 0.0],

                "use_deform": True,

                "joint_limits": {
                    "upper": {
                        "use_limit_x": True,
                        "min_x": -0.523,
                        "max_x": 1.832,

                        "use_limit_y": True,
                        "min_y": 0.0,
                        "max_y": 0.0,

                        "use_limit_z": True,
                        "min_z": 0.0,
                        "max_z": 0.0
                    },

                    "lower": {
                        "use_limit_x": True,
                        "min_x": 0.0,
                        "max_x": 1.571
                    }
                }
            },

            "Leg.L": {
                "prefix": "Leg",
                "side_sign": -1,
                "suffix": "L",

                "parent_bone": "Torso",
                "deform_parent": "Torso",
                "mechanical_parent": "Torso",
                "target_parent": "Torso",
                "pole_parent": "Torso",

                "ik_space": "WORLD",
                "constraint_space": "WORLD",

                "origin": {
                    "region": "hip",
                    "x_factor": 0.65,
                    "z_factor": 0.15
                },

                "segment_length_factor": 1.0,

                "upper_vector": "leg_upper",
                "lower_vector": "leg_lower",

                "pole_vector": "leg_pole",

                "preferred_ik_angle": {
                    "x": -0.1,
                    "y": 0.0,
                    "z": 0.0
                },

                "pole_angle_deg": 0.0,

                "align_roll_axis": [-1.0, 0.0, 0.0],

                "use_deform": True,

                "joint_limits": {
                    "upper": {
                        "use_limit_x": True,
                        "min_x": -0.523,
                        "max_x": 1.832,

                        "use_limit_y": True,
                        "min_y": 0.0,
                        "max_y": 0.0,

                        "use_limit_z": True,
                        "min_z": 0.0,
                        "max_z": 0.0
                    },

                    "lower": {
                        "use_limit_x": True,
                        "min_x": 0.0,
                        "max_x": 1.571
                    }
                }
            }
        },

        "ik_joint_limits": {
            "Arm.Upper.R": {"use_limit_x": True, "min_x": -0.785, "max_x": 1.571, "use_limit_y": True, "min_y": -0.523, "max_y": 0.523, "use_limit_z": True, "min_z": -0.785, "max_z": 1.571},
            "Arm.Upper.L": {"use_limit_x": True, "min_x": -0.785, "max_x": 1.571, "use_limit_y": True, "min_y": -0.523, "max_y": 0.523, "use_limit_z": True, "min_z": -1.571, "max_z": 0.785},
            "Arm.Lower.R": {"use_limit_x": True, "min_x": 0.0, "max_x": 2.094, "use_limit_y": False, "use_limit_z": False},
            "Arm.Lower.L": {"use_limit_x": True, "min_x": 0.0, "max_x": 2.094, "use_limit_y": False, "use_limit_z": False},
            "Leg.Upper.R": {"use_limit_x": True, "min_x": -0.523, "max_x": 1.832, "use_limit_y": True, "min_y": 0.0, "max_y": 0.0, "use_limit_z": True, "min_z": 0.0, "max_z": 0.0},
            "Leg.Upper.L": {"use_limit_x": True, "min_x": -0.523, "max_x": 1.832, "use_limit_y": True, "min_y": 0.0, "max_y": 0.0, "use_limit_z": True, "min_z": 0.0, "max_z": 0.0},
            "Leg.Lower.R": {"use_limit_x": True, "min_x": 0.0, "max_x": 1.571, "use_limit_y": True, "min_y": 0.0, "max_y": 0.0, "use_limit_z": True, "min_z": 0.0, "max_z": 0.0},
            "Leg.Lower.L": {"use_limit_x": True, "min_x": 0.0, "max_x": 1.571, "use_limit_y": True, "min_y": 0.0, "max_y": 0.0, "use_limit_z": True, "min_z": 0.0, "max_z": 0.0}
        },
        "poses": [
            {
                "pose_name": "sitting_on_ground",
                "targets": {
                    "Torso": [0.0, -0.15, 0.38],         # Dropped Z to lower hip to ground, slightly back on Y
                    "Arm.Target.R": [1.1, -0.5, 0.6],
                    "Arm.Target.L": [-1.1, -0.5, 0.6],
                    "Leg.Target.R": [0.25, -0.65, 0.05],  # Brought back along Y so the knees drop flat
                    "Leg.Target.L": [-0.25, -0.65, 0.05], # Kept Z near 0 so feet rest on floor
                    "Leg.Pole.R": [0.25, -0.4, 0.8],
                    "Leg.Pole.L": [-0.25, -0.4, 0.8]
                },
                "right_leg":{
                    "thigh_pitch": 0,
                    "knee_bend": 0,
                    "ankle_pitch": 0
                }
            },
            {
                "pose_name": "standing_right_arm_raised",
                "targets": {
                    "Torso": [0.0, 0.0, 0.95],
                    "Arm.Target.R": [1.1, -0.2, 1.4],
                    "Arm.Target.L": [-1.0, -0.2, 0.7],
                    "Leg.Target.R": [0.2048, -0.05, 0.0],
                    "Leg.Target.L": [-0.2048, -0.05, 0.0],
                    "Arm.Pole.R": [1.2, -0.8, 1.4],
                    "Arm.Pole.L": [-1.0, -1.0, 0.7],
                    "Leg.Pole.R": [0.2048, -2.0, 0.0],
                    "Leg.Pole.L": [-0.2048, -2.0, 0.0]
                },
                "right_leg":{
                    "thigh_pitch": 90,
                    "knee_bend": 90,
                    "ankle_pitch": 0
                }
            }
        ]
    }

    os.makedirs(BASE_DIR, exist_ok=True)
    if os.path.exists(IK_FILE):
        try:
            with open(IK_FILE, "r") as f:
                return json.load(f)
        except Exception:
            pass

    with open(IK_FILE, "w") as f:
        json.dump(default_ik, f, indent=4)

    return default_ik

def production_clear_armature_cache():
    if bpy.context.object and bpy.context.object.mode != 'OBJECT':
        bpy.ops.object.mode_set(mode='OBJECT')

    for obj_name in ["TeddyBear", "TeddyRig"]:
        if obj_name in bpy.data.objects:
            obj = bpy.data.objects[obj_name]
            bpy.context.collection.objects.unlink(obj)
            bpy.data.objects.remove(obj, do_unlink=True)

    for mesh in bpy.data.meshes:
        if "TeddyBearMesh" in mesh.name:
            bpy.data.meshes.remove(mesh)

    for arm in bpy.data.armatures:
        if "TeddyArmatureData" in arm.name:
            bpy.data.armatures.remove(arm)

def purge_initial_scene():
    bpy.ops.wm.read_factory_settings(use_empty=True)

# ==========================================
# 2. MATHEMATICAL MESH GENERATION
# ==========================================

def sign_pow(val, exponent):
    """Preserves sign mapping during non-integer superellipsoid transformations."""
    return (1.0 if val >= 0 else -1.0) * (abs(val) ** exponent)

def generate_superellipsoid(rx, ry, rz, n, subdivisions=3, taper_factor=0.0):
    """Generates an accurate mathematical superellipsoid mesh layout using Mathutils Vectors."""
    t = (1.0 + math.sqrt(5.0)) / 2.0
    base_verts = [
        (-1, t, 0), (1, t, 0), (-1, -t, 0), (1, -t, 0),
        (0, -1, t), (0, 1, t), (0, -1, -t), (0, 1, -t),
        (t, 0, -1), (t, 0, 1), (-t, 0, -1), (-t, 0, 1)
    ]
    faces = [
        (0, 11, 5), (0, 5, 1), (0, 1, 7), (0, 7, 10), (0, 10, 11),
        (1, 5, 9), (5, 11, 4), (11, 10, 2), (10, 7, 6), (7, 1, 8),
        (3, 9, 4), (3, 4, 2), (3, 2, 6), (3, 6, 8), (3, 8, 9),
        (4, 9, 5), (2, 4, 11), (6, 2, 10), (8, 6, 7), (9, 8, 1)
    ]

    proj_verts = [mathutils.Vector(v).normalized() for v in base_verts]

    for _ in range(subdivisions):
        midpoints = {}
        new_faces = []
        def get_midpoint(p1, p2):
            key = tuple(sorted((p1, p2)))
            if key in midpoints:
                return midpoints[key]
            v1, v2 = proj_verts[p1], proj_verts[p2]
            middle = ((v1 + v2) / 2.0).normalized()
            proj_verts.append(middle)
            idx = len(proj_verts) - 1
            midpoints[key] = idx
            return idx

        for tri in faces:
            a = get_midpoint(tri[0], tri[1])
            b = get_midpoint(tri[1], tri[2])
            c = get_midpoint(tri[2], tri[0])
            new_faces.extend([(tri[0], a, c), (tri[1], b, a), (tri[2], c, b), (a, b, c)])
        faces = new_faces

    m = 2.0 / n
    final_verts = []
    for v in proj_verts:
        phi = math.atan2(v.y, v.x)
        theta = math.asin(v.z)

        x = rx * sign_pow(math.cos(theta), m) * sign_pow(math.cos(phi), m)
        y = ry * sign_pow(math.cos(theta), m) * sign_pow(math.sin(phi), m)
        z = rz * sign_pow(math.sin(theta), m)

        if taper_factor != 0.0:
            scale = 1.0 - (taper_factor * v.z)
            x *= scale
            y *= scale

        # FIX: Wrap coordinates in a mathutils.Vector object to expose .x, .y, .z attributes downstream
        final_verts.append(mathutils.Vector((x, y, z)))

    return final_verts, faces


def build_teddy_rig(config, ik_settings):
    # 1. Cleanup previous rig iterations safely
    if "TeddyArmature" in bpy.data.objects:
        bpy.data.objects.remove(bpy.data.objects["TeddyArmature"], do_unlink=True)
    if "TeddyArmature" in bpy.data.armatures:
        bpy.data.armatures.remove(bpy.data.armatures["TeddyArmature"], do_unlink=True)

    # 2. Allocate core data structures
    armature_data = bpy.data.armatures.new("TeddyArmature")
    armature_obj = bpy.data.objects.new("TeddyArmature", armature_data)
    bpy.context.collection.objects.link(armature_obj)

    bpy.context.view_layer.objects.active = armature_obj
    bpy.ops.object.mode_set(mode='EDIT')

    # 3. Extract normalized proportional parameters
    base_r = config.get("base_unit_head_radius", 0.45)
    props = config.get("proportions", {})

    torso_h = base_r * props.get("torso_height_factor", 2.2)
    head_r = base_r
    arm_seg = (base_r * props.get("arm_length_factor", 1.2)) / 2.0
    leg_seg = (base_r * props.get("leg_length_factor", 1.2)) / 2.0

    width_mod = props.get("torso_width_factor", 1.0)
    shoulder_x = base_r * width_mod * 0.85
    shoulder_z = torso_h * 0.75
    hip_x = base_r * width_mod * 0.65
    hip_z = torso_h * 0.15

    ox, oy, oz = config.get("origin", [0.0, 0.0, 0.0])
    bones = armature_data.edit_bones

    # 4. Generate Central Spinal Column
    root = bones.new("Root")
    root.head = (ox, oy, oz)
    #root_len = get_cfg(config, "rig.root_stub_length", 0.1)
    root_len = torso_h
    root.tail = (ox, oy, oz + root_len)

    torso = bones.new("Torso")
    torso.head = (ox, oy, oz + 0.1)
    torso.tail = (ox, oy, oz + 0.1 + torso_h)
    torso.parent = root

    head = bones.new("Head")
    head.head = torso.tail
    head.tail = (ox, oy, torso.tail[2] + head_r * 2.0)
    head.parent = torso

    bone_registry = {
        "Root": root,
        "Torso": torso,
        "Head": head
    }


    # 5. Generic Asset-Agnostic Limb Builder Function
    def build_isolated_limb_chain(
        limb_data,
        origin_joint,
        segment_length,
        bones,
        bone_registry):

        prefix = limb_data["prefix"]
        side_sign = limb_data["side_sign"]
        suffix = limb_data["suffix"]

        deform_parent = bone_registry[
            limb_data["deform_parent"]
        ]

        mechanical_parent = bone_registry[
            limb_data["mechanical_parent"]
        ]

        target_parent = bone_registry[
            limb_data["target_parent"]
        ]

        def_u = f"{prefix}.Upper.{suffix}"
        def_l = f"{prefix}.Lower.{suffix}"

        mch_u = f"Mch.{prefix}.Upper.{suffix}"
        mch_l = f"Mch.{prefix}.Lower.{suffix}"

        t_name = f"{prefix}.Target.{suffix}"

        upper_offset = make_scaled_vector(
          config,
          limb_data["upper_vector"],
          segment_length
        )

        lower_offset = lower_offset = make_scaled_vector(
          config,
          limb_data["lower_vector"],
          segment_length
        )

        du = bones.new(def_u)
        du.head = origin_joint
        du.tail = mathutils.Vector(origin_joint) + upper_offset
        du.parent = deform_parent
        du.use_deform = True

        dl = bones.new(def_l)
        dl.head = du.tail
        dl.tail = du.tail + lower_offset
        dl.parent = du
        dl.use_deform = True

        mu = bones.new(mch_u)
        mu.head = origin_joint
        mu.tail = du.tail.copy()
        mu.parent = mechanical_parent
        mu.use_deform = False

        ml = bones.new(mch_l)
        ml.head = mu.tail
        ml.tail = dl.tail.copy()
        ml.parent = mu
        ml.use_deform = False

        target = bones.new(t_name)
        target.head = ml.tail.copy()
        target.tail = target.head + get_offset(config, "target_tail")
        target.parent = target_parent
        target.use_deform = False

        for b in [du, dl, mu, ml, target]:
            bone_registry[b.name] = b

        return {
            "config": limb_data,
            "deform_upper": du.name,
            "deform_lower": dl.name,
            "mch_upper": mu.name,
            "mch_lower": ml.name,
            "target": target.name
        }


    # 6. Instantiate Appendages from Loaded Parameters
    limb_meta = []

    limb_matrix = ik_settings.get("limb_matrix")


    from pprint import pprint

    print("LIMB MATRIX FOUND:")
    pprint(limb_matrix)


    if limb_matrix is None:
        raise ValueError(
            "Configuration missing required key: 'limb_matrix'"
        )

    if not isinstance(limb_matrix, dict):
        raise TypeError(
            "'limb_matrix' must be a dictionary."
        )

    required_fields = {
        "prefix",
        "side_sign",
        "suffix",
        "deform_parent",
        "mechanical_parent",
        "target_parent",
        "pole_parent",
        "pole_vector"
    }

    for limb_name, limb_cfg in limb_matrix.items():

        missing = required_fields - limb_cfg.keys()

        if missing:
            raise ValueError(
                f"limb_matrix['{limb_name}'] missing: "
                f"{sorted(missing)}"
            )

    for limb_key, data in limb_matrix.items():

        prefix = data["prefix"]
        sign = data["side_sign"]

        if prefix == "Arm":

            origin = (
                ox + shoulder_x * sign,
                oy,
                shoulder_z
            )

            length = arm_seg

        else:

            origin = (
                ox + hip_x * sign,
                oy,
                hip_z
            )

            length = leg_seg

        limb_meta.append(
            build_isolated_limb_chain(
                data,
                origin,
                length,
                bones,
                bone_registry
            )
        )

    # 7. Generate Pole Targets Natively
    for limb in limb_meta:

        get_cfg = limb["config"]

        prefix = get_cfg["prefix"]
        side = get_cfg["suffix"]

        pole_name = f"{prefix}.Pole.{side}"

        pole = bones.new(pole_name)

        lower_mch = bone_registry[
            limb["mch_lower"]
        ]

        #pole.head = (
        #    lower_mch.tail +
        #    get_vector(config, cfg["pole_vector"])
        #)


        if prefix == "Arm":



            segment_length = arm_seg

        else:

            segment_length = leg_seg


        pole.head = (
          lower_mch.head +
          make_scaled_vector(
              config,
              get_cfg["pole_vector"],
              segment_length
          )
        )

        pole.tail = pole.head + get_offset(config,"pole_tail")


        pole.parent = bone_registry[
            get_cfg["pole_parent"]
        ]

        pole.use_deform = False

        bone_registry[pole.name] = pole

    # 8. Switch to Pose Mode to link systems together via constraint targets
    bpy.ops.object.mode_set(mode='POSE')

    for limb in limb_meta:

        limb_cfg = limb["config"]

        def_u = limb["deform_upper"]
        def_l = limb["deform_lower"]

        mch_u = limb["mch_upper"]
        mch_l = limb["mch_lower"]

        t_name = limb["target"]

        p_def_u = armature_obj.pose.bones.get(def_u)
        p_def_l = armature_obj.pose.bones.get(def_l)

        p_mch_u = armature_obj.pose.bones.get(mch_u)
        p_mch_l = armature_obj.pose.bones.get(mch_l)

        if p_mch_u:
            p_mch_u.ik_stretch = 0.0

        if p_mch_l:
            p_mch_l.ik_stretch = 0.0

            ik = p_mch_l.constraints.new('IK')
            ik.target = armature_obj
            ik.subtarget = t_name

            ik.pole_target = armature_obj
            ik.pole_subtarget = f"{limb_cfg['prefix']}.Pole.{limb_cfg['suffix']}"

            #ik_cfg = get_cfg(ik_settings, "ik.default", {})
            ik_cfg = ik_settings.get("ik.default")


            #ik.chain_count = get_cfg(ik_cfg, "chain_count", 2)
            ik.chain_count = ik_cfg["chain_count"]
            #ik.use_rotation = get_cfg(ik_cfg, "use_rotation", False)
            ik.use_rotation = ik_cfg["use_rotation"]

            ik.target_space = limb_cfg.get(
                "ik_target_space",
                "WORLD"
            )

            ik.owner_space = limb_cfg.get(
                "ik_owner_space",
                "WORLD"
            )

            #ik.pole_angle = math.radians(get_cfg(ik_settings, "ik.pole.default_angle_deg", 0.0))
            ik.pole_angle = math.radians(
               limb_cfg.get("pole_angle_deg", 0.0)
            )

        if p_def_u:
            cr = p_def_u.constraints.new('COPY_ROTATION')
            cr.target = armature_obj
            cr.subtarget = mch_u

        if p_def_l:
            cr = p_def_l.constraints.new('COPY_ROTATION')
            cr.target = armature_obj
            cr.subtarget = mch_l




    # 9. Clear initialization local space offsets and clamp lateral hip drift
    #for bone in armature_obj.pose.bones:
    #    if "Target" in bone.name:
    #        edit_bone = armature_data.edit_bones.get(bone.name)
    #        if edit_bone:
    #            bone.location = edit_bone.head - bone.bone.head_local
    #            if "Leg.Target" in bone.name or "Leg.Pole" in bone.name:
    #                bone.location[0] = 0.0
    #
    #    if "Leg.Pole" in bone.name:
    #        bone.location[0] = 0.0




    # ==========================================================================
    # --- DYNAMIC JOINT LIMITS PARSING ---
    # ==========================================================================
    ik_limits = ik_settings.get("ik_joint_limits", {})

    #for limb_key, data in limb_matrix.items():
    #    prefix = data["prefix"]
    #    side = data["suffix"]

    #    for segment_key in ["Upper", "Lower"]:
    #        base_bone_name = f"{prefix}.{segment_key}.{side}"
    #        limits = ik_limits.get(base_bone_name, {})

    #        for target_bone_name in [base_bone_name, f"Mch.{base_bone_name}"]:
    #            p_bone = armature_obj.pose.bones.get(target_bone_name)
    #            if p_bone and limits:
    #                if "use_limit_x" in limits:
    #                    p_bone.use_ik_limit_x = limits["use_limit_x"]
    #                    p_bone.ik_min_x = limits.get("min_x", 0.0)
    #                    p_bone.ik_max_x = limits.get("max_x", 0.0)
    #                if "use_limit_y" in limits:
    #                    p_bone.use_ik_limit_y = limits["use_limit_y"]
    #                    p_bone.ik_min_y = limits.get("min_y", 0.0)
    #                    p_bone.ik_max_y = limits.get("max_y", 0.0)
    #                if "use_limit_z" in limits:
    #                    p_bone.use_ik_limit_z = limits["use_limit_z"]
    #                    p_bone.ik_min_z = limits.get("min_z", 0.0)
    #                   p_bone.ik_max_z = limits.get("max_z", 0.0)

    # 10. Clean exit context handling
    bpy.ops.object.mode_set(mode='OBJECT')
    armature_obj.update_from_editmode()

    return armature_obj

# The rest of your mesh building pipeline, keyframes execution, and diagnostics follows...

def build_teddy_mesh(config, rig_obj=None):
    import bpy
    import math
    import mathutils

    props = config["proportions"]
    hr = config["base_unit_head_radius"]
    exponents = config.get("shape_exponents", {})
    snout_n = exponents.get("snout_boxiness_n", 2.5)
    geom = get_cfg(config, "geometry.superellipsoid", {})

    default_n = get_cfg(geom, "default_n", 2.0)
    snout_n = get_cfg(geom, "snout_n", 2.5)

    taper = get_cfg(config, "geometry.taper.torso", 0.3)


    if not rig_obj:
        rig_obj = bpy.data.objects.get("TeddyArmature")
    if not rig_obj:
        raise RuntimeError("Armature reference missing or 'TeddyArmature' object not found.")

    # Ensure Blender updates its animation and pose matrices before we sample them
    bpy.context.view_layer.update()

    mesh_data = bpy.data.meshes.new("TeddyBearMesh")
    teddy_object = bpy.data.objects.new("TeddyBear", mesh_data)
    bpy.context.collection.objects.link(teddy_object)

    all_verts = []
    all_faces = []
    vertex_groups = {}

    def add_geometry_to_global_arrays(verts, faces, group_name):
        start_idx = len(all_verts)
        all_verts.extend(verts)
        for face in faces:
            all_faces.append(tuple(idx + start_idx for idx in face))
        if group_name not in vertex_groups:
            vertex_groups[group_name] = []
        vertex_groups[group_name].extend(range(start_idx, len(all_verts)))

    def build_bone_aligned_limb(bone_name, rad_start, rad_end, length_scale=1.0):
        # FIXED: Sample the live Pose Bone instead of the static edit bone data
        p_bone = rig_obj.pose.bones.get(bone_name)
        if not p_bone: return

        # Extract world-space coordinates from the pose matrix
        bone_head = p_bone.matrix.to_translation()
        # The bone's local Y axis always points down its length in Blender
        bone_vector = p_bone.matrix.to_quaternion() @ mathutils.Vector((0, p_bone.length, 0))
        bone_length = p_bone.length * length_scale

        v_seg, f_seg = generate_superellipsoid(1.0, 1.0, 1.0, default_n, subdivisions=2)
        transformed_verts = []

        root_extension = 0.0
        if "Arm.Upper" in bone_name or "Leg.Upper" in bone_name:
            root_extension = bone_length * 0.20

        for v in v_seg:
            pct = (v.z + 1.0) / 2.0
            current_r = rad_start + (rad_end - rad_start) * pct
            z_pos = -root_extension + (bone_length + root_extension) * pct

            local_pos = mathutils.Vector((v.x * current_r, v.y * current_r, z_pos))
            rotation_matrix = mathutils.Vector((0, 0, 1)).rotation_difference(bone_vector.normalized()).to_matrix().to_4x4()
            transformed_verts.append((rotation_matrix @ local_pos) + bone_head)

        add_geometry_to_global_arrays(transformed_verts, f_seg, bone_name)

    # ==========================================
    # POSE-ALIGNED CORE BODY & HEAD PIPELINE
    # ==========================================
    p_torso = rig_obj.pose.bones.get("Torso")
    p_head = rig_obj.pose.bones.get("Head")

    if p_torso:
        t_head = p_torso.matrix.to_translation()
        t_vector = p_torso.matrix.to_quaternion() @ mathutils.Vector((0, p_torso.length, 0))
        t_len = p_torso.length

        v_torso, f_torso = generate_superellipsoid(
            hr * props.get("torso_width_factor", 1.4) * 1.25,
            hr * props.get("torso_width_factor", 1.4) * 1.10,
            t_len * 0.5,
            default_n, subdivisions=3, taper_factor=taper
        )
        torso_center = t_head + (t_vector * 0.5)
        v_torso_trans = [v + torso_center for v in v_torso]
        add_geometry_to_global_arrays(v_torso_trans, f_torso, "Torso")

    if p_head:
        h_head = p_head.matrix.to_translation()
        h_vector = p_head.matrix.to_quaternion() @ mathutils.Vector((0, p_head.length, 0))

        v_head, f_head = generate_superellipsoid(hr * 1.1, hr * 1.0, hr * 1.0, default_n, subdivisions=3)
        head_center = h_head + (h_vector * 0.4)
        v_head_trans = [v + head_center for v in v_head]
        add_geometry_to_global_arrays(v_head_trans, f_head, "Head")

        # Head-Relative Face Elements
        v_snout, f_snout = generate_superellipsoid(hr * 0.45, hr * 0.35, hr * 0.3, snout_n, subdivisions=2)
        snout_offset = get_offset(
            config,
            "snout",
            hr
        )

        v_snout_trans = [
            v + head_center + snout_offset
            for v in v_snout
        ]


        #v_snout_trans = [v + head_center + mathutils.Vector((0.0, -hr * 0.85, -hr * 0.1)) for v in v_snout]
        add_geometry_to_global_arrays(v_snout_trans, f_snout, "Head")

        v_nose, f_nose = generate_superellipsoid(hr * 0.15, hr * 0.1, hr * 0.08, 1.8, subdivisions=2)

        nose_offset = get_offset(
            config,
            "nose",
            hr
        )

        v_nose_trans = [
            v + head_center + nose_offset
            for v in v_nose
        ]



        #v_nose_trans = [v + head_center + mathutils.Vector((0.0, -hr * 1.2, hr * 0.08)) for v in v_nose]
        add_geometry_to_global_arrays(v_nose_trans, f_nose, "Head")

        ear_ang = math.radians(props.get("ear_angle_deg", 60.0))
        ear_rad = hr * props.get("ear_radius_factor", 0.35)
        for side_sign in [1, -1]:
            v_ear, f_ear = generate_superellipsoid(ear_rad, ear_rad * 0.3, ear_rad, default_n, subdivisions=2)
            ear_offset = mathutils.Vector((math.sin(ear_ang) * hr * 1.0 * side_sign, -hr * 0.05, math.cos(ear_ang) * hr * 1.0))
            rot_ear = mathutils.Matrix.Rotation(ear_ang * -side_sign, 4, 'Y')
            v_ear_trans = [(rot_ear @ v) + head_center + ear_offset for v in v_ear]
            add_geometry_to_global_arrays(v_ear_trans, f_ear, "Head")

    # --- Generate Arms using absolute Pose Space Alignment ---
    for side in ["R", "L"]:
        build_bone_aligned_limb(f"Arm.Upper.{side}", hr * props["arm_thick_shoulder"], hr * 0.32)
        build_bone_aligned_limb(f"Arm.Lower.{side}", hr * 0.32, hr * props["arm_thick_paw"])

    # --- Generate Legs & Lower Segments (Aligned straight down in Rest Pose) ---
    for side in ["R", "L"]:
        build_bone_aligned_limb(f"Leg.Upper.{side}", hr * 0.45, hr * 0.35)
        build_bone_aligned_limb(f"Leg.Lower.{side}", hr * 0.35, hr * props.get("leg_thick_foot", 0.38))

        # ==========================================================================
        # FOOT OVERRIDE: Inject 90-Degree Forward Rotation
        # ==========================================================================
        foot_bone_name = f"Foot.{side}"
        p_foot_bone = rig_obj.pose.bones.get(foot_bone_name)
        if p_foot_bone:
            # 1. Grab the base translation coordinate where the lower leg meets the foot
            bone_head = p_foot_bone.matrix.to_translation()
            foot_length = p_foot_bone.length * 0.5

            # 2. Generate pristine upright superellipsoid vertices
            v_seg, f_seg = generate_superellipsoid(1.0, 1.0, 1.0, default_n, subdivisions=2)
            transformed_verts = []

            # 3. Create a clean 90-degree rotation matrix around the local X-axis
            rotation_90_deg_forward = mathutils.Matrix.Rotation(get_rotation(config, "foot_forward"), 4, 'X')

            rad_start = hr * props.get("leg_thick_foot", 0.38)
            rad_end = hr * 0.4

            for v in v_seg:
                pct = (v.z + 1.0) / 2.0
                current_r = rad_start + (rad_end - rad_start) * pct
                z_pos = foot_length * pct

                # Scale the raw ellipsoid coordinates
                local_pos = mathutils.Vector((v.x * current_r, v.y * current_r, z_pos))

                # Apply the explicit 90-degree rotation before translating it to the bone space
                rotated_pos = rotation_90_deg_forward @ local_pos

                # Rotate by the bone's global footprint and add to the world location joint head
                transformed_verts.append((p_foot_bone.matrix.to_quaternion() @ rotated_pos) + bone_head)

            add_geometry_to_global_arrays(transformed_verts, f_seg, foot_bone_name)

    mesh_data.from_pydata(all_verts, [], all_faces)
    mesh_data.update()

    for g_name, v_indices in vertex_groups.items():
        v_group = teddy_object.vertex_groups.new(name=g_name)
        v_group.add(v_indices, 1.0, 'REPLACE')

    return teddy_object

def apply_plush_surface_nodes(mesh_obj, config):
    surf = config["surface_deformation"]
    if not surf["use_plush_displacement"]:
        return

    # Modernized for Blender 5.1 geometry modifier node logic
    geom_mod = mesh_obj.modifiers.new(name="ProceduralPlushFuzz", type='NODES')
    node_tree = geom_mod.node_group
    if not node_tree:
        return

    # Python vertex mapping modifier fallback architecture via native displacement stack
    mesh_obj.modifiers.remove(geom_mod)

    tex = bpy.data.textures.new(name="PlushFuzzNoise", type='VORONOI')
    tex.noise_scale = 1.0 / surf["noise_frequency"]

    disp_mod = mesh_obj.modifiers.new(name="PlushFuzz", type='DISPLACE')
    disp_mod.texture = tex
    disp_mod.texture_coords = 'LOCAL'
    disp_mod.strength = surf["noise_scale"]

# ==========================================
# 3. RIGGING ENGINE WITH BIOMECHANICAL IK LIMITS
# ==========================================




def bind_mesh_to_rig(mesh_obj, armature_obj):
    modifier = mesh_obj.modifiers.new(name="TeddyArmatureBinding", type='ARMATURE')
    modifier.object = armature_obj
    modifier.use_vertex_groups = True

# ==========================================
# 4. ENVIRONMENT AND STAGING ENGINE
# ==========================================

def setup_rendering_environment(config, angle_degrees=0.0):
    ox, oy, oz = config["origin"]
    hr = config["base_unit_head_radius"]

    # Calculate target spatial center based on torso height factors
    props = config["proportions"]
    t_height = hr * 2.0 * 1.2 * props.get("torso_height_factor", 1.8)
    target_center = mathutils.Vector((ox, oy, oz + (t_height / 2.0)))

    angle_rad = math.radians(angle_degrees)
    camera_radius = get_cfg(config, "lighting.camera.radius", 5.5)
    cam_x = ox + camera_radius * math.sin(angle_rad)
    cam_y = oy - camera_radius * math.cos(angle_rad)
    cam_z = oz + t_height * get_cfg(config, "lighting.camera.height_factor", 1.1)

    # Clean up previous camera and lighting objects completely
    for obj in list(bpy.data.objects):
        if obj.name in ["RenderCamera", "KeyLight", "FillLight", "RimLight", "FloorLight"]:
            bpy.data.objects.remove(obj, do_unlink=True)

    # 1. CAMERA CONFIGURATION
    camera_data = bpy.data.cameras.new("RenderCameraData")
    camera_obj = bpy.data.objects.new("RenderCamera", camera_data)
    bpy.context.collection.objects.link(camera_obj)

    camera_obj.location = (cam_x, cam_y, cam_z)
    camera_obj.rotation_mode = 'QUATERNION'
    camera_obj.rotation_quaternion = (target_center - camera_obj.location).to_track_quat('-Z', 'Y')
    bpy.context.scene.camera = camera_obj

    # 2. STUDIO AREA LIGHTING ENGINE (Bright, consistent wraparound)
    def create_soft_area_light(name, location, energy, size):
        light_data = bpy.data.lights.new(name=name, type='AREA')
        light_data.shape = 'SQUARE'
        light_data.size = size
        light_data.energy = energy

        # Soften shadow falloff natively
        if hasattr(light_data, 'shadow_soft_size'):
            light_data.shadow_soft_size = size

        light_obj = bpy.data.objects.new(name=name, object_data=light_data)
        bpy.context.collection.objects.link(light_obj)
        light_obj.location = location
        light_obj.rotation_mode = 'QUATERNION'
        light_obj.rotation_quaternion = (target_center - light_obj.location).to_track_quat('-Z', 'Y')
        return light_obj

    # Key Light - Broad, bright primary front-three-quarters lighting
    create_soft_area_light(
        name="KeyLight",
        location=(ox + 3.5 * math.sin(angle_rad + 0.5), oy - 3.5 * math.cos(angle_rad + 0.5), oz + t_height * 2.0),
        energy = get_cfg(config, "lighting.key_light.energy", 150.0),
        size=get_cfg(config, "lighting.key_light.size", 2.0)
    )

    # Fill Light - High ambient bounce on the opposite side to lift all shadow pockets
    create_soft_area_light(
        name="FillLight",
        location=(ox + 4.0 * math.sin(angle_rad - 1.0), oy - 4.0 * math.cos(angle_rad - 1.0), oz + t_height * 0.8),
        energy= get_cfg(config, "lighting.fill_light.energy", 90.0),
        size=get_cfg(config, "lighting.key_light.size", 3.0)
    )

    create_soft_area_light(
        name="FloorLight",
        location=(ox, oy, oz - 1.5),
        energy = get_cfg(config, "lighting.floor_light.energy", 40.0),
        size=get_cfg(config, "lighting.key_light.size", 4.0)
    )

    # 3. ENGINE PIPELINE RESOLUTION
    rc = config["render_settings"]
    bpy.context.scene.render.engine = rc["engine"]
    bpy.context.scene.render.resolution_x = rc["resolution_x"]
    bpy.context.scene.render.resolution_y = rc["resolution_y"]





# ==========================================
# 5. KEYFRAME & ADVANCED DIAGNOSTIC PIPELINE
# ==========================================
def build_keyframes_from_poses(
    armature_obj,
    pose_list,
    ik_settings,
    aconfig):

    if armature_obj.animation_data:
        armature_obj.animation_data_clear()

    print("\n" + "="*60)
    print("[DIAGNOSTIC] --- BEGINNING KEYFRAME MATRIX ANALYSIS ---")
    print("="*60)

    bpy.ops.object.mode_set(mode='POSE')

    anim = ik_settings.get("animation", {})
    rc = aconfig["animation"]

    start_frame = anim.get("start_frame", 1)
    hold = anim.get("hold_frames", rc["hold_frames"])
    transition = anim.get("transition_frames", rc["transition_frames"])

    current_frame = start_frame

    right_leg_definition = build_leg_definition(armature_obj,"R")
    left_leg_definition = build_leg_definition(armature_obj,"L")

    print_leg_definition("RIGHT", right_leg_definition)
    print_leg_definition("LEFT", left_leg_definition)

    for idx, pose_data in enumerate(pose_list):

        pose_name = pose_data.get(
            "pose_name",
            f"Pose_{idx}"
        )

        print(f"\nProcessing {pose_name}")

        controls = build_rig_controls(pose_data)

        #
        # Reset pose
        #

        for bone in armature_obj.pose.bones:
            bone.location = (0,0,0)
            bone.rotation_quaternion = (1,0,0,0)

        bpy.context.view_layer.update()

        apply_torso_controls(
            armature_obj,
            controls
        )

        apply_arm_controls(
            armature_obj,
            "R",
            controls.right_arm
        )

        apply_arm_controls(
            armature_obj,
            "L",
            controls.left_arm
        )

        apply_leg_controls(
            armature_obj,
            "R",
            controls.right_leg,
            right_leg_definition
        )

        apply_leg_controls(
            armature_obj,
            "L",
            controls.left_leg,
            left_leg_definition
        )


        #
        # Keyframe everything
        #

        for bone_name in [
            "Torso",

            "Arm.Target.R",
            "Arm.Target.L",

            "Arm.Pole.R",
            "Arm.Pole.L",

            "Leg.Target.R",
            "Leg.Target.L",

            "Leg.Pole.R",
            "Leg.Pole.L"
        ]:

            keyframe_bone(
                armature_obj,
                bone_name,
                current_frame
            )

        current_frame += hold

        for bone_name in [
            "Torso",

            "Arm.Target.R",
            "Arm.Target.L",

            "Arm.Pole.R",
            "Arm.Pole.L",

            "Leg.Target.R",
            "Leg.Target.L",

            "Leg.Pole.R",
            "Leg.Pole.L"
        ]:

            keyframe_bone(
                armature_obj,
                bone_name,
                current_frame
            )

        current_frame += transition

    bpy.ops.object.mode_set(mode='OBJECT')

    return current_frame


def run_config_impact_analysis(mesh_obj, armature_obj, config, ik_settings):
    print("\n" + "="*80)
    print("  CONFIGURATION IMPACT MATRIX ANALYSIS")
    print("="*80)
    props = config["proportions"]
    hr = config["base_unit_head_radius"]
    t_height = hr * 2.0 * 1.2 * props["torso_height_factor"]
    total_arm_len = t_height * props["arm_length_factor"]

    print(f"[*] Calculated Mathematical Torso Height: {t_height:.3f} units")
    print(f"[*] Max Operational Arm Reach Segment Bounds: {total_arm_len:.3f} units")
    print("="*80 + "\n")

def generate_spatial_anchor_registry(armature_obj, mesh_obj, end_frame):
    pass

def apply_diagnostic_materials():
    pass

def configure_video_render_pipeline(end_frame, config):
    rc = config["render_settings"]

    scene = bpy.context.scene
    scene.frame_start = get_cfg(config, "animation.start_frame", 1)
    scene.frame_end = end_frame

    arc =config["animation"]
    scene.render.fps = get_cfg(config, "animation.fps", arc["fps"])

    scene.render.image_settings.file_format = rc.get("image_format", "JPEG")

    scene.render.ffmpeg.format = rc.get("format", "MPEG4")
    scene.render.ffmpeg.codec = rc.get("video_codec", "H264")

    scene.render.filepath = rc.get("output_path", OUTPUT_VIDEO_PATH)

# ==========================================
# 6. MATHEMATICAL AUTOMATED UNIT TESTING ENGINE
# ==========================================

def validate_bear_proportions(mesh_obj, config):
    """
    Unit Test V4: Evaluates true postured bounding box dimensions.
    """
    import bpy
    # Get the evaluated object state from the dependency graph
    depsgraph = bpy.context.evaluated_depsgraph_get()
    eval_mesh_obj = mesh_obj.evaluated_get(depsgraph)

    bbox = [eval_mesh_obj.matrix_world @ mathutils.Vector(corner) for corner in eval_mesh_obj.bound_box]
    min_c = mathutils.Vector((min([v.x for v in bbox]), min([v.y for v in bbox]), min([v.z for v in bbox])))
    max_c = mathutils.Vector((max([v.x for v in bbox]), max([v.y for v in bbox]), max([v.z for v in bbox])))

    total_dimensions = max_c - min_c
    print(f"[Verification Node] True Posed Bounding Box: Width={total_dimensions.x:.2f}m, Depth={total_dimensions.y:.2f}m, Height={total_dimensions.z:.2f}m")

    # Sitting Pose Validation Logic:
    # A sitting bear's bounding depth (Y) must extend forward due to the straight legs.
    if total_dimensions.y < (config["base_unit_head_radius"] * 2.0):
        raise AssertionError(
            "POSE FAULT: Bear is standing or legs failed to project forward into a sitting posture! "
            "Validation terminated."
        )


def validate_overall_bear_shape(mesh_object, rig_object, tolerance=0.02):
    """
    Unit Test V4: Evaluates dynamic deformed spatial continuity relative to rest anatomy.
    """
    import bpy
    print("Executing Evaluated Dependency Graph Permutation Matrix...")

    # Helper to capture vertex world centers for a given rig evaluation mode ('POSE' or 'REST')
    def capture_vertex_centers(eval_mode):
        orig_mode = rig_object.data.pose_position
        rig_object.data.pose_position = eval_mode
        bpy.context.view_layer.update()

        depsgraph = bpy.context.evaluated_depsgraph_get()
        eval_obj = mesh_object.evaluated_get(depsgraph)
        deformed_mesh = eval_obj.to_mesh()

        group_lookup = {g.index: g.name for g in mesh_object.vertex_groups}
        centers = {}

        for vertex in deformed_mesh.vertices:
            for g in vertex.groups:
                g_name = group_lookup.get(g.group)
                if g_name:
                    if g_name not in centers:
                        centers[g_name] = []
                    centers[g_name].append(eval_obj.matrix_world @ vertex.co)

        eval_obj.to_mesh_clear()
        rig_object.data.pose_position = orig_mode  # Restore original state safely
        return centers

    # 1. Gather pristine reference data from REST pose and current active data from POSE mode
    rest_centers = capture_vertex_centers('REST')
    pose_centers = capture_vertex_centers('POSE')

    structural_permutation_tree = [
        ("Torso", "Head"),
        ("Torso", "Arm.Upper.R"), ("Torso", "Arm.Upper.L"),
        ("Torso", "Leg.Upper.R"), ("Torso", "Leg.Upper.L"),
        ("Arm.Upper.R", "Arm.Lower.R"), ("Arm.Upper.L", "Arm.Lower.L"),
        ("Leg.Upper.R", "Leg.Lower.R"), ("Leg.Upper.L", "Leg.Lower.L")
    ]

    for parent_name, child_name in structural_permutation_tree:
        # Pull rest state geometry arrays
        p_verts_rest = rest_centers.get(parent_name, [])
        c_verts_rest = rest_centers.get(child_name, [])

        # Pull posed state geometry arrays
        p_verts_pose = pose_centers.get(parent_name, [])
        c_verts_pose = pose_centers.get(child_name, [])

        if not p_verts_rest or not c_verts_rest or not p_verts_pose or not c_verts_pose:
            raise ValueError(f"Topology Array Missing Data: {parent_name} or {child_name}")

        # Compute the natural distance separating these vertex groups at Rest
        min_rest_dist = float('inf')
        for p_v in p_verts_rest:
            for c_v in c_verts_rest:
                d = (p_v - c_v).length
                if d < min_rest_dist: min_rest_dist = d

        # Compute the actual current distance separating them in the active Pose
        min_pose_dist = float('inf')
        for p_v in p_verts_pose:
            for c_v in c_verts_pose:
                d = (p_v - c_v).length
                if d < min_pose_dist: min_pose_dist = d

        # The true indicator of joint tearing is the difference between rest and pose states


        distance_deviation = abs(min_pose_dist - min_rest_dist)
        print(f"  - Evaluated Link [{parent_name:11} <-> {child_name:11}]: Separation Shift = {distance_deviation:.4f}m (Rest: {min_rest_dist:.3f}m, Pose: {min_pose_dist:.3f}m)")

        # DYNAMIC TOLERANCE ENFORCEMENT:
        # Give joints that compress heavily (like the hips) a slightly wider tolerance window
        active_tolerance = (
            cfg(config, "validation.tolerances.hip_joint", tolerance)
            if ("Leg" in child_name or "Torso" in parent_name)
            else tolerance
        )

        if distance_deviation > active_tolerance:
            raise AssertionError(
                f"DEFORMATION BREAKAGE: Joint separation caught during posturing between '{parent_name}' and '{child_name}'. "
                f"Limb mesh stretched by {distance_deviation:.4f}m beyond its rest design dimensions. Build halted."
            )

    print("Validation Successful: Posed asset contains perfect structural integrity relative to rest geometry.")
    return True

def verify_rendered_posture_and_silhouette(mesh_obj, rig_obj, config):
    """
    Unit Test V7: Strict Anatomical Bone Axis Audit (Object Space).
    Verifies that the upper leg bones point forward along the object's Y axis.
    """
    import bpy
    import math
    import mathutils

    print("\n" + "="*60)
    print("[Verification Node] Running True Bone-Axis Pose Audit...")
    print("="*60)

    # Ensure Blender updates the runtime pose positions
    bpy.context.view_layer.update()

    # In Blender Object Space, the spine vector points straight UP (+Z)
    object_spine_vertical = mathutils.Vector((0.0, 0.0, 1.0))

    angles = []

    for side in ["R", "L"]:
        p_bone = rig_obj.pose.bones.get(f"Leg.Upper.{side}")
        if p_bone:
            # Head (Hip joint) and Tail (Knee joint) locations in pure object space
            head_pos = p_bone.head
            tail_pos = p_bone.tail

            # The actual spatial direction vector of the thigh bone in the scene
            bone_direction = (tail_pos - head_pos).normalized()

            # Calculate the angle between the physical bone path and the spine
            dot_product = bone_direction.dot(object_spine_vertical)
            angle_rad = math.acos(max(-1.0, min(1.0, dot_product)))
            angle_deg = math.degrees(angle_rad)
            angles.append(angle_deg)

            print(f"[*] Thigh Physical Axis Analysis ({side}):")
            print(f"  - Head (Hip): ({head_pos.x:.3f}, {head_pos.y:.3f}, {head_pos.z:.3f})")
            print(f"  - Tail (Knee): ({tail_pos.x:.3f}, {tail_pos.y:.3f}, {tail_pos.z:.3f})")
            print(f"  - True Direction Vector: ({bone_direction.x:.4f}, {bone_direction.y:.4f}, {bone_direction.z:.4f})")
            print(f"  - Measured Angle to Spine: {angle_deg:.2f}°")

    if not angles:
        raise AssertionError("PIPELINE FAULT: Thigh bones could not be evaluated.")

    avg_angle = sum(angles) / len(angles)
    print("-" * 60)
    print(f"[*] Evaluated Posture Angle: {avg_angle:.2f}°")
    print("-" * 60)

    # Assert true sitting posture (Thighs perpendicular to spine: 70° to 110°)
    min_a, max_a = cfg(config, "validation.pose_angle_range_deg", [65.0, 115.0])

    if avg_angle < min_a or avg_angle > max_a:
        raise AssertionError(
            f"POSTURE REJECTION: Physical leg bones are hanging down at {avg_angle:.2f}°. "
            f"They fail to project forward horizontally (Target: ~90°). Build halted."
        )

    print("SUCCESS: True sitting posture confirmed by bone-axis verification. Proceeding to render pipeline.")
    print("="*60 + "\n")
    return True

def validate_rig_foundation(rig_obj, config):
        print("--- Running Rig Foundation Audit ---")
        base_r = config["base_unit_head_radius"]
        props = config["proportions"]

        # 1. Expected Math Calculation
        expected_torso_h = base_r * props.get("torso_height_factor", 1.8)
        expected_arm_len = (base_r * props.get("arm_length_factor", 1.2))

        # 2. Get the pose bones
        bpy.ops.object.mode_set(mode='POSE')

        def check_dist(bone_name, expected_len):
            bone = rig_obj.pose.bones.get(bone_name)
            if not bone: return False
            # Calculate bone length in world space
            actual_len = bone.length
            # Tolerance of 0.05 units for floating point variance
            if abs(actual_len - expected_len) > 0.05:
                raise AssertionError(f"Rig Foundation Error: {bone_name} length {actual_len:.3f} does not match expected {expected_len:.3f}")
            return True

        # 3. Perform Audit
        check_dist("Torso", expected_torso_h)
        check_dist("Arm.Upper.R", expected_arm_len / 2.0)

        # 4. Location Audit: Ensure shoulder is at expected offset
        shoulder = rig_obj.pose.bones.get("Arm.Upper.R")
        torso = rig_obj.pose.bones.get("Torso")
        if shoulder and torso:
            # Check if shoulder is roughly above the torso origin (simple height check)
            if shoulder.head.z < torso.head.z:
                raise AssertionError("Rig Foundation Error: Shoulder joint is below torso base!")

        print("--- Rig Foundation Audit Passed ---")
        bpy.ops.object.mode_set(mode='OBJECT')

def validate_solver_readiness(rig_obj):
    # Check if the bone is ACTUALLY at the origin in 3D space
    for bone in rig_obj.pose.bones:
        if "Target" in bone.name:
            world_pos = bone.matrix.to_translation()
            if world_pos.length < 0.1:
                raise AssertionError(f"Target {bone.name} is at origin! IK will pull limbs to center.")
    print("Solver Readiness Audit Passed.")






# ==========================================
# 7. EXECUTION DRIVER
# ==========================================

if __name__ == "__main__":




    # ==========================================================================
    # GLOBAL CLEAN COLLECTION SWEEP: EVAPORATE BLENDER DEFAULT STARTUP ASSETS
    # ==========================================================================
    # Clean default objects (Cube, Camera, Light) to avoid enclosing the bear
    for obj in list(bpy.data.objects):
        # Target the default cube directly, or sweep all generic startup meshes
        if obj.name in ["Cube", "Camera", "Light"] or obj.type == 'MESH' and obj.name == "Cube":
            bpy.data.objects.remove(obj, do_unlink=True)

    # Sweep target structural meshes from prior programmatic script loops
    if "TeddyBearMesh" in bpy.data.objects:
        bpy.data.objects.remove(bpy.data.objects["TeddyBearMesh"], do_unlink=True)
    if "TeddyBearMesh" in bpy.data.meshes:
        bpy.data.meshes.remove(bpy.data.meshes["TeddyBearMesh"], do_unlink=True)

    # Load runtime parameters and baseline tracking constraints
    system_config = load_or_create_config()
    ik_parameters = load_or_create_ik_defaults()

    target_angle = system_config.get("proportions", {}).get("ear_angle_deg", 45.0)

    # ==========================================================================
    # RESTORED INTEGRATED EXECUTION PIPELINE HIERARCHY
    # ==========================================================================

    # 1. Instantiate the skeletal armature layout
    teddy_rig_object = build_teddy_rig(system_config, ik_parameters)

    for side in ["R", "L"]:
        upper = teddy_rig_object.pose.bones[f"Mch.Leg.Upper.{side}"]
        lower = teddy_rig_object.pose.bones[f"Mch.Leg.Lower.{side}"]
        target = teddy_rig_object.pose.bones[f"Leg.Target.{side}"]

        hip = upper.head
        foot_target = target.head

        chain_length = upper.length + lower.length
        target_distance = (foot_target - hip).length

        print(f"\nLeg {side}")
        print(f"  Chain length     : {chain_length:.4f}")
        print(f"  Target distance  : {target_distance:.4f}")
        print(f"  Difference       : {chain_length - target_distance:.4f}")

    validate_rig_foundation(teddy_rig_object, system_config)

    validate_solver_readiness(teddy_rig_object)

    # 2. Compute timeline keyframe tracks first so animation channels are registered
    poses = ik_parameters.get("poses", [])

    rc = system_config["animation"]
    if len(poses) >= 2:
        total_frames = build_keyframes_from_poses(teddy_rig_object, poses, ik_parameters, system_config)
    else:
        total_frames = START_FRAME + rc["hold_frames"] + rc["transition_frames"]

    # 3. FORCE THE ARMATURE SKELETON INTO A CLEAN, UN-DEFORMED REST STATE
    # This prevents the geometry engine from mathematical squashing or baking
    # the arms directly inside the body coordinates.
    teddy_rig_object.data.pose_position = 'REST'

    # ==========================================================================
    # CRITICAL MATRIX FLUSH FIX:
    # Force Blender to completely update its internal dependency graph and
    # recalculate the bone matrices back to the straight REST position.
    # ==========================================================================
    bpy.context.view_layer.update()
    bpy.context.evaluated_depsgraph_get()


    # 4. Generate the un-deformed geometric mesh over pristine T-pose proportions
    teddy_mesh_object = build_teddy_mesh(system_config, teddy_rig_object)

    # --------------------------------------------------------------------------
    # DIAGNOSTIC TELEMETRY MILESTONE A: MESH TRANSFORMS & APPLIED MATRIX
    # --------------------------------------------------------------------------
    print("\n[DEBUG A] --- CHECKING MESH TRANSFORMS PRE-BIND ---")
    print(f"  - Mesh Location:       {teddy_mesh_object.location}")
    print(f"  - Mesh Scale:          {teddy_mesh_object.scale}")
    print(f"  - Mesh Matrix World:\n{teddy_mesh_object.matrix_world}")
    print(f"  - Rig Matrix World:\n{teddy_rig_object.matrix_world}")
    # --------------------------------------------------------------------------

    # 5. Bind the un-deformed mesh to the rig structure via Armature Modifiers
    bind_mesh_to_rig(teddy_mesh_object, teddy_rig_object)

    # --------------------------------------------------------------------------
    # DIAGNOSTIC TELEMETRY MILESTONE B: VERTEX WEIGHT ISOLATION
    # --------------------------------------------------------------------------
    print("\n[DEBUG B] --- VERTEX GROUP ASSIGNMENT CHECK ---")
    vgroups = {vg.name: vg.index for vg in teddy_mesh_object.vertex_groups}
    print(f"  - Vertex Groups Found: {list(vgroups.keys())}")

    # Check if Arm.Upper.R actually has vertices assigned to it
    if "Arm.Upper.R" in vgroups:
        idx = vgroups["Arm.Upper.R"]
        weighted_verts = [v for v in teddy_mesh_object.data.vertices if any(g.group == idx for g in v.groups)]
        print(f"  - Vertices assigned to 'Arm.Upper.R': {len(weighted_verts)}")
    else:
        print("  - WARNING: 'Arm.Upper.R' vertex group does not exist on the mesh!")
    # --------------------------------------------------------------------------

    # 6. RESTORE THE SKELETON TO POSE MODE SO THE TIMELINE EVALUATES DYNAMICALLY
    teddy_rig_object.data.pose_position = 'POSE'
    bpy.context.scene.frame_set(START_FRAME)

    # Flush matrices again so the upcoming validation runs on the posed bear
    bpy.context.view_layer.update()
    dg = bpy.context.evaluated_depsgraph_get()

    # --------------------------------------------------------------------------
    # DIAGNOSTIC TELEMETRY MILESTONE C: POSE BONE VS EDIT BONE COORDINATES
    # --------------------------------------------------------------------------
    print("\n[DEBUG C] --- SPACE DISCONNECT MATRIX AUDIT ---")
    torso_pose = teddy_rig_object.pose.bones.get("Torso")
    arm_pose = teddy_rig_object.pose.bones.get("Arm.Upper.R")

    if torso_pose and arm_pose:
        # Get evaluated world space positions directly from the matrices
        torso_world_tail = (teddy_rig_object.matrix_world @ torso_pose.matrix).to_translation()
        arm_world_head = (teddy_rig_object.matrix_world @ arm_pose.matrix).to_translation()
        math_dist = (torso_world_tail - arm_world_head).length

        print(f"  - Torso Pose Tail (World): {torso_world_tail}")
        print(f"  - Arm.Upper.R Head (World): {arm_world_head}")
        print(f"  - Mathematical Bone Gap in Pose Mode: {math_dist:.4f}m")
        print(f"  - Arm.Upper.R Constraints: {[c.type for c in arm_pose.constraints]}")
    # --------------------------------------------------------------------------

    # Before calling validate_overall_bear_shape(...)
    teddy_mesh_object.update_tag()
    dg.update()
    bpy.context.view_layer.update()

    # --------------------------------------------------------------------------
    # DIAGNOSTIC TELEMETRY MILESTONE D: EVALUATED MESH GEOMETRY
    # --------------------------------------------------------------------------
    print("\n[DEBUG D] --- DEPSGRAPH EVALUATED MESH AUDIT ---")
    eval_mesh_obj = teddy_mesh_object.evaluated_get(dg)
    temp_mesh = eval_mesh_obj.to_mesh()
    print(f"  - Evaluated Mesh Vertex Count: {len(temp_mesh.vertices)}")
    eval_mesh_obj.to_mesh_clear()
    print("--------------------------------------------------\n")
    # --------------------------------------------------------------------------

    # Force a hard frame recalculation to bake constraint matrices
    current_frame = bpy.context.scene.frame_current

    # Move to next frame and back to flush the evaluation stack
    bpy.context.scene.frame_set(current_frame + 1)
    bpy.context.scene.frame_set(current_frame)

    # Force a final graph sync
    dg = bpy.context.evaluated_depsgraph_get()
    dg.update()
    bpy.context.view_layer.update()

    # Manually compute and apply constraint matrices to the pose bones
    bpy.ops.object.mode_set(mode='POSE')
    dg = bpy.context.evaluated_depsgraph_get()

    # Force evaluation of bone matrices
    for bone in teddy_rig_object.pose.bones:
        # This manually updates the bone matrix based on current constraints
        bone.matrix = teddy_rig_object.matrix_world @ bone.matrix

    bpy.ops.object.mode_set(mode='OBJECT')
    bpy.context.view_layer.update()

    # ==========================================================================
    # PIPELINE INTERCEPT FIX: FORCE CORRECT LATERAL LEG TARGET ALIGNMENT
    # ==========================================================================
    #print("[PIPELINE INTERCEPT] Intercepting frame constraints to clamp lateral leg split...")
    #bpy.ops.object.mode_set(mode='POSE')

    # 0.2048 matches the exact spatial coordinate of the hip sockets from Telemetry
    #hip_offset_x = 0.2048

    #target_r = teddy_rig_object.pose.bones.get("Leg.Target.R")
    #target_l = teddy_rig_object.pose.bones.get("Leg.Target.L")
    #pole_r   = teddy_rig_object.pose.bones.get("Leg.Pole.R")
    #pole_l   = teddy_rig_object.pose.bones.get("Leg.Pole.L")

    # Force coordinates inward to prevent side-splitting leg rotation
    #if target_r: target_r.location[0] = hip_offset_x
    #if target_l: target_l.location[0] = -hip_offset_x
    #if pole_r:   pole_r.location[0]   = hip_offset_x
    #if pole_l:   pole_l.location[0]   = -hip_offset_x

    # Push update back into the graph evaluation queue
    #bpy.context.view_layer.update()
    #bpy.ops.object.mode_set(mode='OBJECT')

    # --------------------------------------------------------------------------
    # RELEVANT DIAGNOSTIC TELEMETRY: LEG IK SOLVER MISALIGNMENT AUDIT
    # --------------------------------------------------------------------------
    print("\n==================================================")
    print("[DEBUG LEG TWIST] --- ISOLATING IK PLANE DATA ---")
    print("==================================================")

    bpy.ops.object.mode_set(mode='POSE')

    # Target bones to inspect
    leg_names = ["Mch.Leg.Upper.R", "Mch.Leg.Lower.R", "Leg.Target.R", "Leg.Pole.R"]

    for bname in leg_names:
        p_bone = teddy_rig_object.pose.bones.get(bname)
        if p_bone:
            # 1. Extract evaluated world positions
            world_pos = (teddy_rig_object.matrix_world @ p_bone.matrix).to_translation()
            print(f"\nBone: {bname}")
            print(f"  - World Position:         {world_pos}")

            # 2. Extract Edit Mode structural orientation properties
            e_bone = teddy_rig_object.data.edit_bones.get(bname) if hasattr(teddy_rig_object.data, "edit_bones") else None
            # If not in edit mode right now, we pull from the bone data block directly
            b_data = teddy_rig_object.data.bones.get(bname)

            if b_data:
                print(f"  - Vector (Head to Tail):  {(b_data.tail_local - b_data.head_local).normalized()}")

            # 3. Check Pose constraints and execution influence metrics
            if p_bone.constraints:
                for c in p_bone.constraints:
                    if c.type == 'IK':
                        print(f"  - CONSTRAINT FOUND:       IK")
                        print(f"    * Pole Angle Assigned:  {c.pole_angle} radians ({(c.pole_angle * 180 / 3.14159):.1f}°)")
                        print(f"    * Chain Count:          {c.chain_count}")
                        print(f"    * Tracked Subtarget:    {c.subtarget}")
                        print(f"    * Pole Subtarget:       {c.pole_subtarget}")

            # 4. Read the Local Rotation Matrix to identify twist anomalies
            loc, rot, scale = p_bone.matrix.decompose()
            euler_rot = rot.to_euler()
            print(f"  - Evaluated Local Euler:  X: {euler_rot.x:.2f}, Y: {euler_rot.y:.2f}, Z: {euler_rot.z:.2f} (Radians)")

    bpy.ops.object.mode_set(mode='OBJECT')
    print("==================================================\n")


    # --------------------------------------------------------------------------
    # DIAGNOSTIC TELEMETRY MILESTONE E: RUNTIME CONSTRAINT EVALUATION CRASH AUDIT
    # --------------------------------------------------------------------------
    print("\n==================================================")
    print("[DEBUG SOLVER] --- RUNTIME MATRIX EVALUATION ---")
    print("==================================================")
    bpy.ops.object.mode_set(mode='POSE')

    for side in ["R", "L"]:
        mch_lower_name = f"Mch.Leg.Lower.{side}"
        deform_lower_name = f"Leg.Lower.{side}"
        target_name = f"Leg.Target.{side}"

        p_mch = teddy_rig_object.pose.bones.get(mch_lower_name)
        p_def = teddy_rig_object.pose.bones.get(deform_lower_name)
        p_targ = teddy_rig_object.pose.bones.get(target_name)

        if p_mch and p_def and p_targ:
            mch_world_tail = (teddy_rig_object.matrix_world @ p_mch.matrix).to_translation()
            targ_world_head = (teddy_rig_object.matrix_world @ p_targ.matrix).to_translation()
            distance_to_target = (mch_world_tail - targ_world_head).length

            print(f"\nLimb Side [{side}]:")
            print(f"  - Target World Coordinate: [{targ_world_head.x:.4f}, {targ_world_head.y:.4f}, {targ_world_head.z:.4f}]")
            print(f"  - Actual Foot End Position: [{mch_world_tail.x:.4f}, {mch_world_tail.y:.4f}, {mch_world_tail.z:.4f}]")
            print(f"  - Solver Distance Error:    {distance_to_target:.6f}m")

            # Read out rotation values to check if bone is completely locked
            print(f"  - Deform Bone Local Euler:  X: {p_def.rotation_euler.x:.2f}, Y: {p_def.rotation_euler.y:.2f}")
            if p_mch.constraints:
                for c in p_mch.constraints:
                    if c.type == 'IK':
                        print(f"  - IK Constraint Active:     {not c.mute}")
                        print(f"  - IK Influence Metric:     {c.influence}")

    bpy.ops.object.mode_set(mode='OBJECT')
    print("==================================================\n")




    # ==========================================================================
    # POST-VALIDATION AUDITS & RENDER PIPELINE INTERACTIVE EVALS
    # ==========================================================================
    ##validate_overall_bear_shape(teddy_mesh_object, teddy_rig_object, tolerance=0.02)

    # Force the scene context evaluation frame update to capture live postures
    current_frame = bpy.context.scene.frame_current
    bpy.context.scene.frame_set(current_frame)

    for bone in bpy.context.object.data.bones:
       print(bone.name)



    bpy.context.scene.frame_set(1)
    bpy.context.view_layer.update()


    setup_rendering_environment(system_config, angle_degrees=target_angle)
    run_config_impact_analysis(teddy_mesh_object, teddy_rig_object, system_config, ik_parameters)

    # 7. Safe trigger pipeline for the final MP4 video output rendering sequence
    if len(poses) >= 2:
        #configure_video_render_pipeline(total_frames, system_config)
        configure_video_render_pipeline(60, system_config)

        # Explicit hardware render block to write frames automatically out to disk
        print(f"Executing video rendering sequence -> {OUTPUT_VIDEO_PATH}")
        #bpy.ops.render.render(animation=True)


        # Force every scene in the data block
        for scene in bpy.data.scenes:
            try:
                scene.render.engine = 'BLENDER_EEVEE_NEXT'
            except TypeError:
                scene.render.engine = 'BLENDER_EEVEE'

        # Explicitly force the CURRENT context scene that the file saves as active
        try:
            bpy.context.scene.render.engine = 'BLENDER_EEVEE_NEXT'
            print(f"Set context engine: {bpy.context.scene.render.engine}")
        except TypeError:
            bpy.context.scene.render.engine = 'BLENDER_EEVEE'


        # Save the work you just did in the notebook to a temporary blend file
        bpy.ops.wm.save_as_mainfile(filepath="/tmp/notebook_scene.blend")





```
